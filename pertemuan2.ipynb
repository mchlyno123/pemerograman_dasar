```python
# Inisialisasi variabel
pin_count = 16  # int (jumlah pin IC)
operating_voltage = 3.3  # float (tegangan kerja dalam volt)
is_surface_mount = True  # bool (apakah komponen tipe SMD)
component_name = "ATmega328P"  # str (nama komponen)

# Mencetak nilai dan tipe data masing-masing variabel
print(pin_count, type(pin_count))
print(operating_voltage, type(operating_voltage))
print(is_surface_mount, type(is_surface_mount))
print(component_name, type(component_name))
```

```python
# Inisialisasi variabel dengan tipe data yang sesuai
unit_name = "PLTS Nunukan"  # str (teks/string)
panel_count = 240  # int (bilangan bulat)
output_voltage = 48.5  # float (bilangan desimal)
is_operating = True  # bool (nilai kebenaran: True/False)

# Mencetak nilai beserta tipe datanya
print(unit_name, type(unit_name))
print(panel_count, type(panel_count))
print(output_voltage, type(output_voltage))
print(is_operating, type(is_operating))
```

```python
tegangan_harian = [217.5, 219.0, 220.3, 218.7, 221.1, 216.9]

# (a) Cetak pembacaan pertama dan terakhir memakai indeks
print("Pembacaan pertama:", tegangan_harian[0])
print("Pembacaan terakhir:", tegangan_harian[-1])

# (b) Tambahkan satu pembacaan baru 222.0
tegangan_harian.append(222.0)

# (c) Cetak jumlah total data dengan len()
print("Jumlah total data:", len(tegangan_harian))
```

```python
# Membuat dictionary spesifikasi motor listrik
motor_spec = {
    "nama": "Motor Induksi 3-Phasa",
    "daya_kW": 15.0,
    "tegangan_V": 380,
    "rpm": 1450,
}

# 1. Cetak salah satu nilai lewat key-nya (contoh: daya_kW)
print("Daya motor:", motor_spec["daya_kW"], "kW")

# 2. Ubah nilai rpm menggunakan operator =
motor_spec["rpm"] = 1500

# 3. Cetak ulang seluruh dictionary
print("Spesifikasi terbaru:", motor_spec)
```

```python
# Daftar tiga unit pembangkit (List of Dictionary)
pembangkit = [
    {"nama": "Unit-1", "kapasitas_MW": 100, "status": "operasi"},
    {"nama": "Unit-2", "kapasitas_MW": 150, "status": "operasi"},
    {"nama": "Unit-3", "kapasitas_MW": 80, "status": "pemeliharaan"},
]

# 1. Cetak nama unit kedua (indeks 1)
print("Nama unit kedua:", pembangkit[1]["nama"])

# 2. Ubah status salah satu unit (contoh: ubah status Unit-3 menjadi "operasi")
pembangkit[2]["status"] = "operasi"

# 3. Tambahkan satu unit baru ke list dengan append()
pembangkit.append({"nama": "Unit-4", "kapasitas_MW": 120, "status": "standby"})

# 4. Cetak seluruh data
print("\nSeluruh data pembangkit:")
for unit in pembangkit:
    print(unit)
```
